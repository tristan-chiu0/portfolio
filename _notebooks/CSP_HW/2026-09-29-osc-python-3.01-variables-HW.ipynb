## Design Thinking

**Player Need:** Some players may be uncomfortable or face serious health effects such as epilepsy from fast movement or animation.

**Goal:** Give the player freedom over full motion or reduced motion while keeping the game's functionality

**Representation Considered:** A boolean is useful for reduced motion because the setting is simply either On or Off (True or False). Numerical variables work well for Star counts as those are numerical values.

**Prototype:** The star counter stores three star counters for each level, a total score, and the reduced motion boolean setting.

**Test/Revision:** Testing made me realize that the stored total must be recalculated afer each level value changes. Reduced motion should be a player choice as each player differs; Some players may need/prefer it while others may not.
