## Homework Hack (independent): SFI Backend QA Simulator

```python
%%python

# CODE_RUNNER: Homework Hack

# Homework Hack - SFI Backend QA Simulator

import random

parts = [
    {
        "product_name": "Replacement Flywheels",
        "category": "Auto Racing",
        "spec_number": "1.1"
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2"
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1"
    }
]

actions = [
    "GET/search",
    "POST/create",
    "PUT/update",
    "DELETE/remove"
]

existing_spec_numbers = ["1.1", "2.1"]
test_count = 4

# Build and test your QA simulator below.

def run_qa_test(test_number):
    selected_part = random.choice(parts)
    selected_action = random.choice(actions)
    spec = selected_part["spec_number"]
    name = selected_part["product_name"]

    print(f"--- Test {test_number} ---")
    print(f"Selected Part: {name} (Spec {spec}, {selected_part['category']})")
    print(f"Selected Action: {selected_action}")

    if selected_action == "GET/search":
        print(f"Result: Searching backend for '{name}' (Spec {spec})...")

    elif selected_action == "POST/create":
        duplicate_spec = spec in existing_spec_numbers
        if duplicate_spec:
            print(f"Result: REJECTED — Spec {spec} already exists in backend.")
        else:
            print(f"Result: ACCEPTED — Creating new record for '{name}' (Spec {spec}).")

    elif selected_action == "PUT/update":
        print(f"Result: Updating existing record for '{name}' (Spec {spec})...")

    elif selected_action == "DELETE/remove":
        print(f"Result: Removing record for '{name}' (Spec {spec}) from backend...")

    print()


for i in range(1, test_count + 1):
    run_qa_test(i)
```
