## Popcorn Hack - Fill in the Blanks
1. The **checkbox** on a toggle keeps track of whether it is on or off.
2. The **slider** is the visible part of the toggle that changes color when it is turned on.
3. The .switch class controls the toggles position, size, padding, and **alignment**.
4. The **border-radius** Property makes the slider have rounded edges.
5. The &before class dictates the **white-circular portion** of the toggle.


## Homework Hack - Interactive Toggle Panel
This solution incorporates three OCS toggles that control different features on the page. Each toggle has its own purpose, and JavaScript updates the page based on the toggle's status.

// Clear the output
outputElement.innerHTML = "";

// Create the interactive toggle panel using OCS classes
outputElement.innerHTML =
  '<div class="ocs__container">' +
  '<h2 class="ocs__section-title">Interactive Toggle Panel</h2>' +
  '<p class="ocs__description">Try each toggle and see what happens!</p>' +

    '<span class="ocs__toggle-track" aria-hidden="true"></span>' +
    '<span class="ocs__toggle-label">Reveal Creature</span>' +
  '</label>' +
  '<img id="bird1" src="{{ site.baseurl }}/images/about/bird.png" alt="A bird" hidden>' +
  '<p id="image-status">The creature is hidden.</p>' +

  '<label class="ocs__toggle ocs__toggle--green">' +
    '<input id="message-toggle" class="ocs__toggle-input" type="checkbox">' +
    '<span class="ocs__toggle-track" aria-hidden="true"></span>' +
    '<span class="ocs__toggle-label">Show Message</span>' +
  '</label>' +
  '<p id="message" hidden>Great job! You are learning interactive programming!</p>' +

  '<label class="ocs__toggle ocs__toggle--green">' +
    '<input id="dark-toggle" class="ocs__toggle-input" type="checkbox">' +
    '<span class="ocs__toggle-track" aria-hidden="true"></span>' +
    '<span class="ocs__toggle-label">Show Dark Mode Message</span>' +
  '</label>' +
  '<p id="dark-status" hidden>Dark mode would be enabled!</p>' +

  '<h3>Toggle Statistics</h3>' +
  '<p>Number of toggles turned on: <strong id="toggle-count">0</strong> / 3</p>' +

// Select each toggle
const imageToggle = outputElement.querySelector("#image-toggle");
const messageToggle = outputElement.querySelector("#message-toggle");
const darkToggle = outputElement.querySelector("#dark-toggle");

// Select the elements controlled by the toggles
const image = outputElement.querySelector("#bird1");
const imageStatus = outputElement.querySelector("#image-status");
const message = outputElement.querySelector("#message");
const darkStatus = outputElement.querySelector("#dark-status");
const countDisplay = outputElement.querySelector("#toggle-count");

// This function counts how many toggles are currently checked.
// I learned that .checked returns a Boolean value: true or false.
function updateToggleCount() {
  const toggles = outputElement.querySelectorAll(".ocs__toggle-input");
  let count = 0;

  toggles.forEach((toggle) => {
    if (toggle.checked) {
      count++;
    }
  });

  countDisplay.textContent = count;
}

// Toggle 1: Reveal or hide the creature image
imageToggle.addEventListener("change", () => {
  image.hidden = !imageToggle.checked;

  if (imageToggle.checked) {
    imageStatus.textContent = "The creature is visible!";
  } else {
    imageStatus.textContent = "The creature is hidden.";
  }

  updateToggleCount();
});

// Toggle 2: Show or hide a message
messageToggle.addEventListener("change", () => {
  message.hidden = !messageToggle.checked;

  // The ! operator reverses the Boolean value.
  // This lets the same line handle both showing and hiding.
  updateToggleCount();
});

// Toggle 3: Show or hide a dark mode message
darkToggle.addEventListener("change", () => {
  darkStatus.hidden = !darkToggle.checked;

  updateToggleCount();
});
