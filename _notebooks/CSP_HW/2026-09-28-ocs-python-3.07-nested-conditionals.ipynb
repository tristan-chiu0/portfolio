## Popcorn Hack 1: Predicting Outcome
Prediction: Since is_weekday = True then the if statement homework_finished will be ran. Since home_finished is False, the code print("Finish homework") will be printed

```python
%%python

# CODE_RUNNER: Popcorn Hack 1 Prediction

is_weekday = True
homework_finished = False

if is_weekday:
    if homework_finished:
        print("Free time")
    else:
        print("Finish homework")
else:
    print("Weekend plan")
```

## Homework Hack: High School Daily Routine Dashboard

```python
%%python

# CODE_RUNNER: Homework Hack 

student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True
}

# Write your complete nested conditional below.
# Do not delete the starter data. Test several value combinations.
if student["logged_in"]:
    if student["homework_finished"]:
        if student["lunch_packed"]:
            if student["practice_today"]:
                print("nothing to do today!")
            else:
                print("Go to practice today")
        else:
            print("Need to pack lunch")
    else:
        print("Need to finish homework.")
else:
    print("Please log in to access your decision maker.")
```
