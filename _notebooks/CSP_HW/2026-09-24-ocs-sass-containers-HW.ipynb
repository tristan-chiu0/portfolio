## OCS Container Grammar and HW
The finished HW and Popcorn hacks are below, utilizing UI Runners. Click Run on each labeled UI runner to preview the layout. Both solutions use the shared OCS container grammar with no custom CSS. ## Homework Hack - Poway Rotary Club 
The parade hub uses one container and one card in an organized manner. It includes the required standard cols-3 grid with 3 cells, an OCS table with two body rows, and a final callout.

{% capture challenge_b %}
Homework Hack Challenge - SASS containers for Parade Page
{% endcapture %}
{% capture code_b %}
// Clear the output
outputElement.innerHTML = "";

// Inject semantic HTML
outputElement.innerHTML = 
    "<div class='ocs__container'>" +
    "<h2 class='ocs__section-title'>Poway Rotary Parade</h2>" +

    "<div class='ocs__card'>" +
      "<div class='ocs__grid--standard cols-3'>" +
      "<div class='ocs__grid-cell ocs__grid-cell--header'>Parade Info</div>" +
      "<div class='ocs__grid-cell'>Location: Poway Road</div>" +

      "</div>" +
    "</div>" +

      "<table class='ocs__table'>" +
